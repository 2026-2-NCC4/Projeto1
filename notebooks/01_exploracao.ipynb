**LIMPANDO E UNIFORMIZANDO DADOS**

```python
contas_selecionadas = [
    "BAL - Ativo Circulante",
    "BAL - Disponível",
    "BAL - Estoques Diversos",
    "BAL - Realizável a Longo Prazo",
    "BAL - Passivo Circulante",
    "BAL - Exigível a Longo Prazo",
    "BAL - Patrimônio Líquido",
    "BAL - Permanente",
    "BAL - Emprést",
    "BAL - Empréstimos",
    "DRE - Receita",
    "DRE - Tributos",
    "DRE - Custos",
    "DRE - Resultado Operacional",
    "DRE - Resultado Antes do Imposto de Renda",
    "DRE - Imposto de Renda e Contribuição Social",
    "DRE - Resultado Líquido",
    "DRE - Receitas Financeiras",
    "DRE - Despesas Financeiras",
    "DRE - EBITDA",
    "FLU - Geração de Caixa",
    "FLU - Investimentos",
    "FLU - Saldo Inicial",
    "FLU - Saldo Final"
]

contas_encontradas = (
    df[df["Conta"].isin(contas_selecionadas)]["Conta"]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)

pd.DataFrame({
    "Contas encontradas": contas_encontradas
})
#RESULTADO: varias contas nao apreceram porque possuem nome diferente, como por exemplo "BAL - Empréstimos" e "BAL - Emprést", que são a mesma conta, mas com nomes diferentes. Isso mostra que a base de dados não está totalmente padronizada,
```

```python
df["Conta"] = (
    df["Conta"]
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)
#uniformizando os nomes das contas, removendo espaços extras e padronizando a formatação
```

```python
df[
    df["Conta"].isin(contas_selecionadas)
]["Conta"].drop_duplicates().sort_values().reset_index(drop=True)
#as 24 contas apareceram padronizadas
```

**Limpeza e uniformização dos dados**

A etapa de limpeza foi realizada sobre a base selecionada, considerando as características dos dados financeiros.

Inicialmente, foi realizada a padronização dos nomes das contas, removendo espaços em excesso. Também foi verificada a existência de valores ausentes, não sendo identificados valores nulos na base selecionada.

A coluna `Valor`, originalmente armazenada como texto no formato numérico brasileiro, foi convertida para o tipo `float64`, permitindo a realização de cálculos e análises estatísticas.

Foram identificados 129.150 valores negativos e 2.112 valores iguais a zero. Esses valores foram mantidos, pois podem representar situações financeiras válidas, como custos, despesas, investimentos e ausência de receitas financeiras.

Também foram identificados valores extremos, sendo o menor valor de aproximadamente -R$ 8,89 bilhões e o maior de aproximadamente R$ 14,59 bilhões. Esses valores foram mantidos nesta etapa, pois não foram considerados erros automaticamente e poderão ser avaliados posteriormente durante a análise dos dados.

Por fim, não foram encontradas duplicatas exatas na base selecionada.

Foi observado ainda que quatro contas — BAL - Emprést, BAL - Estoques Diversos, BAL - Exigível a Longo Prazo e BAL - Patrimônio Líquido — não apresentam registros no Ano 12. Essa ausência foi mantida, sem preenchimento artificial, para preservar as características originais da base.
