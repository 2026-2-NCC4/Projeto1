```python
df_amostra = pd.read_csv(
    caminho,
    sep=";",
    encoding="latin-1",
    header=None,
    names=["Ano", "Cenario", "Conta", "Valor"], #mostra as as colunas que tem na base de dados
    nrows=1000
)

df_amostra.head(10)
```

```python
df_amostra.info() #mostra informações sobre a base de dados, como o tipo de cada coluna e se tem valores nulos
#resultado: temos 1000 linhas e 4 colunas, sendo que a coluna Valor é do tipo float64, e as colunas Ano, Cenario e Conta são do tipo object. A coluna Valor tem 43 valores nulos, o que representa 4,3% dos dados. As outras colunas não tem valores nulos.
```

```python
df_amostra["Ano"].unique() #mostra os anos que tem na base de dados nas primeiras 1000 linhas
```

```python
df_amostra["Cenario"].unique() #mostra os cenários que tem na base de dados nas primeiras 1000 linhas
# resultado: dentro das primeiras 1000 linhas, temos 1 ano, 15 cenarios
```

```python
df_amostra["Cenario"].value_counts() #mostra a quantidade de vezes que cada cenário aparece na base de dados nas primeiras 1000 linhas
# resultado: no cenario 00001, tem 71 registros, no cenario 00002, tem 71 registros, e assim por diante. O cenario 00015 tem 70 registros, que so mostrou parcialmente aqui. Isso mostra que a base de dados está balanceada, ou seja, cada cenário tem aproximadamente o mesmo número de registros.
```

```python
df_amostra[df_amostra["Cenario"] == "Total Cen_00001"]["Conta"].unique() #descobrindo quais contas aparecem dentro de uma cenário específico, no caso o cenario 00001. O resultado mostra que dentro do cenario 00001, temos 71 contas diferentes, que são as mesmas contas que aparecem nos outros cenarios, então a base de dados está balanceada.
#resultado: temos 71 contas diferentes, que são as mesmas contas que aparecem nos outros cenarios, então a base de dados está balanceada.
```

```python
df_amostra["Conta"].dropna().str[:3].value_counts() #mostra a quantidade de vezes que cada conta aparece na base de dados nas primeiras 1000 linhas, considerando apenas os 3 primeiros caracteres da conta, que são os que indicam o tipo de conta. 
#df_amostra["Conta"] mostra só a coluna Conta
#.dropna() remove os valores nulos
#.str[:3] pega os 3 primeiros caracteres da conta

# O resultado mostra que temos 4 tipos de contas diferentes: 1. Receitas, 2. Custos, 3. Despesas e 4. Resultado.
# no info () vimos 957 valores preenchidos em conta, entao bateu. ou seja, temos 69 contas diferentes que aparece varias vezes uma vez em cada cenario
```

```python
df_amostra[df_amostra["Cenario"] == "Total Cen_00001"].isna().sum() #diz quantos valores vazios tem em cada coluna de um cenario completo
#resultado, no cenario 00001, temos 0 valores nulos em Ano, Cenario e Conta, e 3 valores nulos em Valor. Isso mostra que a base de dados está balanceada, ou seja, cada cenário tem aproximadamente o mesmo número de registros.
df_amostra[
    (df_amostra["Cenario"] == "Total Cen_00001") &
    (df_amostra["Conta"].isna())
] #mostra os registros onde a conta é nula no cenario 00001
df_amostra["Cenario"].value_counts()
```
